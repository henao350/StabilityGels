```python
from ngsolve import *
import numpy as np
from netgen.csg import *
from netgen.meshing import MeshingParameters, meshsize
from ngsolve.webgui import Draw

### MAIN ###

L = 90.0; w=15; d=1.6;
# maxh = 0.55
maxh = 0.41
order = 1
mesh_filename='meshes/mesh-test'

left  = Plane (Pnt(-0.5*L,0,0), Vec(-1,0,0) )
right = Plane (Pnt(0.5*L,0,0), Vec( 1,0,0) )
back  = Plane (Pnt(0,-0.5*w,0), Vec(0,-1, 0) )
front = Plane (Pnt(0,0.5*w,0), Vec(0,1,0) )
bot   = Plane (Pnt(0,0,0), Vec(0,0,-1) ).bc('bottom')
top   = Plane (Pnt(0,0,d), Vec(0,0,1) )

brick = left * right * front * back * bot * top

geo = CSGeometry()
geo.Add (brick)

mesh = Mesh(geo.GenerateMesh(maxh=maxh))
mesh.ngmesh.Save(mesh_filename)
Draw(mesh)

fes = VectorH1(mesh, order=order, dirichlet="bottom")
print(f'nDoF = {fes.ndof}')
```
