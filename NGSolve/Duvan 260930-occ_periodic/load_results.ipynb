```python
from ngsolve import *
from ngsolve.webgui import Draw
from ClassGel2d import gel2d
import numpy as np

# ---------------------------------------------------------
# Choose the simulation to visualize
# ---------------------------------------------------------
L = 100.0
d = 10.0
order = 3
phi0 = 0.7
mu_input = 0.0029
mu_bar = -mu_input

simulation_name = (
    f'test_periodic_L{int(L)}_d{int(d)}'
    f'_phi0={phi0}_nIterations100'
)

# Select the saved continuation step.
# 0 = first perturbed equilibrium at lambda=2.
# Larger values correspond to larger swelling.
numIteration = 1

print(
    f'L={L} [mm], d={d} [mm], order={order}, '
    f'phi0={phi0}, mu_bar={mu_bar}'
)

gel = gel2d(L, d, phi0, mu_bar)

mesh_file = 'meshes/meshPeriodic.vol'
mesh = Mesh(mesh_file)

fes = Periodic(
    VectorH1(mesh, order=order, dirichlet='bot')
)
print(f'sum(fes.FreeDofs()) = {sum(fes.FreeDofs())}')

filename = (
    f'{simulation_name}/result_gelsMu2D'
    + gel.filename_suffix
    + '_iter='
    + str(numIteration).zfill(2)
)

print(f'filename={filename}.gfu')

u = GridFunction(fes)
u.Load(filename + '.gfu')

F = Id(2) + Grad(u)

energy_density = Integrate(gel.W(F), mesh)/(L*d)
print('Average energy density = ', energy_density)

# ---------------------------------------------------------
# Surface amplitude
# ---------------------------------------------------------
top_values = []

for p in mesh.ngmesh.Points():
    x0, y0 = p[0], p[1]
    if abs(y0-d) < 1e-8:
        top_values.append(float((y+u[1])(mesh(*p))))

if top_values:
    surface_amplitude = 0.5*(max(top_values)-min(top_values))
    print('Surface amplitude = ', surface_amplitude)
    print('Surface amplitude / d = ', surface_amplitude/d)

# Normalized deformed height.
Draw((y+u[1])/d, mesh, deformation=u)
print(' ')
```
