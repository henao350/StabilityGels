```python
from ngsolve import *
from ngsolve.webgui import Draw
from gels2d_ChemicalPotential import gel2d
import numpy as np
# import matplotlib.pyplot as plt
# import scipy.optimize

# data = sys.argv
# In gels2d_ChemicalPotential.py the arguments are passed from shell
# Here, we specify them 'by hand', taking the values from, e.g., the gelsMu?.out screen printout file
# They are, in that order:
# the length; the thickness; the order for the finite element space;
# the initial polymer volume fraction phi0; 
# the absolute value of the nondimensionalized chemical potential
#    mu_bar = mu / (K_B*T)
# (which, in turn, will be assumed to be negative)
# For example,
#    90 1.62 3 0.2 0.5
# means L=90, d=1.62, order=3, phi0=0.2, mu_bar=-0.5

data = ['dummy', '90', '1.62', '3', '0.2', '0.002']
# data = ['dummy', '90', '1.62', '3', '0.2', '0.005']
L = float(data[1])
d = float(data[2])
order = int(data[3])
phi0 = float(data[4])
mu_input = float(data[5])    
mu_bar = -mu_input  # internal sign convention
print(f'L={L} [mm], d={d} [mm], order={order}, phi0={phi0}, mu_bar={mu_bar}')
#
gel = gel2d(L,d,phi0,mu_bar)
#
mesh_file = 'meshes1_62/mesh51.vol'
mesh = Mesh(mesh_file)
# Draw(mesh)
#
# Finite element space with zero-displacement boundary condition on the whole interface
fes = VectorH1(mesh, order=order, dirichlet="bonded_interface|debonded_interface")
print('nDoF = {}'.format(fes.ndof))
#
filename = 'gridfunctions/result_gelsMu2D'+gel.filename_suffix + f'_order={order}'
print(f'filename={filename}')   
#
# u = GridFunction(fes) 
# u.Load(filename + '.gfu')
# F = Id(2) + Grad(u)                                                                                                                         
# # Draw(gel.W(F), mesh, deformation=u)
# Draw(gel.W(F), mesh)
#
nIterations = 15
indexes_iterations = range(nIterations-1,nIterations+1)
for numIteration in indexes_iterations:
    u = GridFunction(fes) 
    filename_with_iter_suffix = filename + '_iter=' + str(numIteration).zfill(2) + '.gfu'
    print(filename_with_iter_suffix)
    u.Load(filename_with_iter_suffix)    
    F = Id(2) + Grad(u)
    print('Average energy density = ', round(Integrate(gel.W(F), mesh)/(L*d),5))   
    Draw((y+u[1])/d, mesh, deformation=u)
    print(' ')

    
```
