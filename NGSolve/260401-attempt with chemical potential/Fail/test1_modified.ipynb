```python
# Nueva función mu_fun (reemplazar con eq. 3.6 de Kang & Juang)
def mu_fun(lambda_val, phi0):
    return - (lambda_val - 1.0) * phi0
```

```python
import numpy as np

def Solve_incremental_mu(phi0, lambda_target=2.0, n_steps=16):
    lambdas = np.linspace(1.1, lambda_target, n_steps)
    mus = [mu_fun(l, phi0) for l in lambdas]

    results = []
    for l, mu in zip(lambdas, mus):
        print(f"Resolviendo para lambda={l:.3f}, mu={mu:.4f}")
        # Aquí iría SolveNonlinearMinProblem con mu fijo
        results.append((l, mu))

    return results
```
