```python
import netgen.geom2d as geom2d
from ngsolve import *
from ngsolve.webgui import Draw

L = 100
d = 20

geo = geom2d.SplineGeometry()
points_list = [(-0.5*L, 0), (0.5*L,0), \
               (0.5*L, d), (-0.5*L,d)]
p1,p2,p3,p4 = [geo.AddPoint(x,y) for x,y in points_list]
geo.Append(["line",p1,p2],bc='bonded')
print("bonded:", [p1,p2])
l_right = geo.Append(["line",p2,p3], bc="right")
print("right:", [p2,p3])
geo.Append(["line",p3,p4])
l_left = geo.Append(["line",p4,p1], bc="left", copy=l_right)
print("left:", [p4,p1])

mesh = Mesh(geo.GenerateMesh(maxh=4))
Draw(mesh)
filename='meshes/mesh.vol'
mesh.ngmesh.Save(filename)

V_std = H1(mesh, order=2, dirichlet="bottom")
V = Periodic(V_std)

print(f"Standard DOFs: {V_std.ndof}")
print(f"Periodic DOFs: {V.ndof}")
```
